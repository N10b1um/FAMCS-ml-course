## Коротко о теории

### 1. Логистическая регрессия
Логистическая регрессия моделирует вероятность положительного класса:

$$
P(y=1 \mid x) = \sigma(z), \quad z = w^T x + b
$$

где $\sigma(z)$ — сигмоида:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

### 2. Почему не линейная регрессия
Если предсказывать класс линейной моделью напрямую, прогноз может выйти за диапазон $[0, 1]$.  
Сигмоида решает эту проблему: она переводит любое число в вероятность.

### 3. Функция потерь
Для логистической регрессии используется **log-loss**:

$$
L = - \frac{1}{n} \sum_{i=1}^{n}
\left[
y_i \log(\hat{p}_i) + (1-y_i)\log(1-\hat{p}_i)
\right]
$$

### 4. Обучение
Мы будем минимизировать log-loss методом **градиентного спуска**.

### 5. Что будем сравнивать
После своей реализации вы сравните результат с `sklearn.linear_model.LogisticRegression`:
- accuracy,

### 6. Почему нужна стандартизация
Градиентный спуск работает стабильнее, когда признаки находятся примерно в одном масштабе.

## Шаг 2. Реализуем сигмоиду

Допишите функцию `sigmoid(z)`.

Проверка:
- `sigmoid(0)` должно быть равно `0.5`;
- функция должна работать и со скалярами, и с `numpy`-массивами.

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print("sigmoid(0) =", sigmoid(0))
print("sigmoid([-1, 0, 1]) =", sigmoid(np.array([-1.0, 0.0, 1.0])))
```

## Шаг 3. Реализуем log-loss

Напишите функцию `compute_log_loss(y_true, y_pred_proba)`.

Важно:
- чтобы избежать `log(0)`, используйте `np.clip(y_pred_proba, 1e-15, 1 - 1e-15)`.

```python
def compute_log_loss(y_true, y_pred_proba):
    y_pred_proba = np.clip(y_pred_proba, 1e-15, 1 - 1e-15)
    loss = -np.mean(y_true * np.log(y_pred_proba) + (1 - y_true) * np.log(1 - y_pred_proba))
    return loss

def l2_penalty(alpha, weights):
    return alpha * np.sum(np.square(weights))

def total_loss(y_true, y_pred_proba, alpha, weights):
    return compute_log_loss(y_true, y_pred_proba) + l2_penalty(alpha, weights)

toy_y = np.array([0, 1, 1, 0])
toy_p = np.array([0.1, 0.9, 0.8, 0.3])
print("Toy log-loss:", compute_log_loss(toy_y, toy_p))
```

## Шаг 4. Один шаг градиентного спуска

Напомним:
- $z = Xw + b$
- $\hat{p} = \sigma(z)$

Градиенты для log-loss:
$$
dw = \frac{1}{n} X^T(\hat{p} - y), \qquad db = \frac{1}{n}\sum(\hat{p} - y)
$$

Ниже нужно реализовать функцию, которая считает:
- вероятности,
- loss,
- градиенты $dw$, $db$.

```python
def forward_backward(X, y, w, b, alpha = 0.01):
    z = X @ w + b;
    p = sigmoid(z)
    loss = total_loss(y, p, alpha, w)

    error = p - np.array(y).reshape(-1)
    dw = (1 / X.shape[0]) * (X.T @ error) + 2 * alpha * w
    db = np.mean(error)
    return p, loss, dw, db

n_features = X_train_scaled.shape[1]
w0 = np.zeros(n_features)
b0 = 0.0

p0, loss0, dw0, db0 = forward_backward(X_train_scaled, y_train.values, w0, b0)
print("Initial loss:", loss0)
print("dw shape:", dw0.shape)
print("db:", db0)
```

## Шаг 5. Собираем модель в класс

Допишите методы:
- `fit`
- `predict_proba`
- `predict`

Требования:
- обучение через градиентный спуск;
- сохранять историю значения loss;
- в `predict` использовать порог `threshold`.

```python
class MyLogisticRegressionGD:
    # alpha выше, чем нужно! нормальное alpha здесь в районе 0.001, но я спецально 
    # поставил высокое значение, чтобы наглядно был виден результат регуляризации.
    # при alpha = 0.001 модель выдает те же ответы, что и без регуляризации.
    def __init__(self, learning_rate=0.05, n_iters=1000, threshold=0.5, alpha=0.01): 
        self.learning_rate = learning_rate
        self.n_iters = n_iters
        self.threshold = threshold
        self.alpha = alpha
        self.w = None
        self.b = None
        self.loss_history = []

    def fit(self, X, y):
        n_features = X.shape[1]
        self.w = np.zeros(n_features)
        self.b = 0.0

        for i in range(self.n_iters):
            p, loss, dw, db = forward_backward(X, y, self.w, self.b, self.alpha)
            self.w = self.w - self.learning_rate * dw
            self.b = self.b - self.learning_rate * db
            self.loss_history.append(loss)
        return self

    def predict_proba(self, X):
        z = X @ self.w + self.b
        proba = sigmoid(z)
        return proba

    def predict(self, X):
        proba = self.predict_proba(X)
        preds = (proba > self.threshold).astype(int)
        return preds
```

## Шаг 6. Обучаем свою модель

Попробуйте обучить свою модель и посмотрите:
- уменьшается ли `loss`;
- какое качество получается на test.

```python
my_model = MyLogisticRegressionGD(
    learning_rate=0.01,   # подберите разумное значение
    n_iters=1000,         # задайте число итераций
    threshold=0.5
)

my_model = my_model.fit(X_train_scaled, y_train)

plt.figure(figsize=(8, 4))
plt.plot(my_model.loss_history)
plt.title("История функции потерь")
plt.xlabel("Итерация")
plt.ylabel("Log-loss")
plt.show()
```

## Шаг 7. Оценка качества своей модели

Заполните код ниже и посчитайте:
- accuracy,
- precision,
- recall,
- f1,
- ROC-AUC.

Подсказка:
- для ROC-AUC нужны **вероятности**, а не классы.

```python
# YOUR CODE HERE
my_proba_test = my_model.predict_proba(X_test_scaled)
my_pred_test = my_model.predict(X_test_scaled)

# Я знаю, что это можно было сделать через sklearn, но я очень хотел потренироваться на метриках ручками. 
tp = np.sum((my_pred_test == 1) & (y_test == 1))
fp = np.sum((my_pred_test == 1) & (y_test == 0))
tn = np.sum((my_pred_test == 0) & (y_test == 0))
fn = np.sum((my_pred_test == 0) & (y_test == 1))

precision = tp / (tp + fp)
recall = tp / (tp + fn)

my_metrics = {
    "accuracy": (tp + tn) / len(y_test),
    "precision": precision,
    "recall": recall,
    "f1": 2 * (precision * recall) / (precision + recall),
    "roc_auc": roc_auc_score(y_score = my_proba_test, y_true = y_test),
}

pd.Series(my_metrics).round(4)
```

## Шаг 8. Confusion matrix и ROC-кривая

```python
fig, ax = plt.subplots(figsize=(5, 5))
ConfusionMatrixDisplay.from_predictions(y_test, my_pred_test, ax=ax)
plt.title("Confusion matrix: MyLogisticRegressionGD")
plt.show()

fig, ax = plt.subplots(figsize=(6, 5))
RocCurveDisplay.from_predictions(y_test, my_proba_test, ax=ax)
plt.title("ROC curve: MyLogisticRegressionGD")
plt.show()

print(classification_report(y_test, my_pred_test))
```

### **Ответы**
1. Цель сигмоиды - сжать предсказания линейной регрессии (которые могут быть любыми) до [0,1], обозначающих уверенность модели в классе.
2. MSE слишком слабо штрафует. Штраф MSE в logreg был бы ограничен единицей. Log Loss может жёстко штрафовать модель за уверенность в ошибке.
3. Стандартизация необходима, чтобы модель не считала признак более важным из-за того, что числа в нём больше. например, модель, работающая с деньгами в тыс. $ игнорировала бы возраст, который исчесляется десятками. Плюс стандартизация повышает скорость сходимости и делает градиентный спуск стабильнее.
4. Модели выдали практически одинаковый результат. Различие связано с тем, что по дефолту logReg от scikitlearn использует L2 регуляризацию и lbfgs вместо градиентного спуска. UPD: после добавления в нашей версии регуляризации, результат остался таким же, но мы смогли понизить веса.
5. Во-первых Sklearn использует в качестве алгоритма не радиантный спуск, а LBFGS. Во-вторых библиотека написана на C.
6. При повышении порога растёт precision, но падает recall. При понижении порога precision падает, а recall растёт. Можно сказать, что порогом мы регулируем строгость модели. При повышении порога у нас меньше ложноположительных классификаций, но и пропускаем мы больше реальнх случаев. При понижении же мы начинаем реже пропускать больных, но чаще классифицируем здоровых как людей с опухолью.
7. 
    Плюсы LogReg:
    - Логистическая регрессия даёт отличную интерпретируемость. Мы можем сказать, что признак важный, если на нём высокий вес.
    - LogReg нормально работает на небольших наборах данных, в то время как другие модели, вероятнее всего, на таком же датасете переобучатся.
    - Часто логистическую регрессию используют как baseline. Если она выдаёт достаточно хорошие результаты, возможно, нет смысла брать более сложные модели ради мизерных процентов.

    Минусы:4
    - Линейность. Модель не видит нелинейные зависимости. Если мы захотим научить модель работать с ними, нам придётся заниматься feature engineering'ом, и создавать функции от признаков (их квадраты, кубы, синусы и т.д.).
    - Она не умеет работать с наборами признаков. Иногда комбинация признаков (прим. индекс массы тела > 30.0 + курение) дают вместе огромный эффект. Логистическая регрессия просто умножит признаки на веса и сложит, а деревья бы увидели такую зависимость. Эту проблему тоже можно решить, создав новый признак, но сама по себе LogReg работает с взаимодействиями хуже, чем деревья.
